```python
import random

# Inspection Station · Module I capstone
LIMIT_LO = 24.95
LIMIT_HI = 25.05
accepted = 0
rejected = 0
consecutive_rejects = 0
alarm_count = 0
first_alarm_part = None

for part in range(1, 201):
    # Add tool-wear drift
    if part <= 100:
        dia = round(random.uniform(24.85, 25.15), 3)
    else:
        dia = round(random.uniform(24.90, 25.20), 3)

    if LIMIT_LO <= dia <= LIMIT_HI:
        accepted += 1
        print(f"Part {part}: {dia} mm -   ACCEPT ✓")
        consecutive_rejects = 0 # Reset on accept
    else:
        rejected += 1
        print(f"Part {part}: {dia} mm -   REJECT ✗")
        consecutive_rejects += 1 # Increment on reject

        # Check for tool-wear alarm
        if consecutive_rejects == 3:
            print("⚠ TOOL WEAR SUSPECTED at part", part)
            alarm_count += 1
            if first_alarm_part is None:
                first_alarm_part = part

print("\nShift report:")
print(f"- Total accepted: {accepted}")
print(f"- Total rejected: {rejected}")
acceptance = round(accepted / 200 * 100, 1)
print(f"- Acceptance percentage: {acceptance}%")
print(f"- Number of tool-wear alarms: {alarm_count}")
print(f"- First alarm part number: {first_alarm_part}")
print(f"Verification: Total parts processed = {accepted + rejected}")
```
