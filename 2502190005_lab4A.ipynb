```python
import random

# Inspection Station · Module I capstone
LIMIT_LO = 24.95
LIMIT_HI = 25.05
accepted = 0
rejected = 0

for part in range(1, 201):
    dia = round(random.uniform(24.85, 25.15), 3)

    # The original check for negative 'dia' is not needed here
    # because random.uniform generates positive numbers within its range.

    if LIMIT_LO <= dia <= LIMIT_HI:
        accepted += 1
        print("  ACCEPT ✓")
    else:
        rejected += 1
        print("  REJECT ✗")

print("Shift report:", accepted, "accepted,", rejected, "rejected")
print(f"Verification: Total parts processed = {accepted + rejected}")
```
