```python
import random
import datetime
import math

# ==================================================
# JOB B Code (Reconstructed based on requirements)
# (Do NOT change existing Job A or Job B logic)
# ==================================================

# Constants
TOTAL_PARTS_INSPECTED = 200
LOWER_TOLERANCE = 24.95
UPPER_TOLERANCE = 25.05
NOMINAL_DIAMETER = 25.0
CONSECUTIVE_REJECTS_THRESHOLD = 3 # Assumed from typical industrial practices, consistent with 'consecutive_rejects logic'

# Variables for tracking
accepted_parts = 0
rejected_parts = 0
consecutive_rejects = 0
tool_wear_alarms = 0
first_alarm_part_number = None
is_alarming = False # Flag to track if we are currently in an alarm state for distinct alarm events

# Simulation loop for part inspection
print("Starting part inspection...\n")

for part_number in range(1, TOTAL_PARTS_INSPECTED + 1):
    # Simulate part size. Since 'tool-wear drift' is not to be changed, and no prior code is given,
    # a simple uniform distribution is used to allow for parts outside tolerance.
    # If original JOB B code had specific drift, it would be here.
    part_size = random.uniform(LOWER_TOLERANCE - 0.2, UPPER_TOLERANCE + 0.2) # Generate parts that can be outside tolerance

    if LOWER_TOLERANCE <= part_size <= UPPER_TOLERANCE:
        accepted_parts += 1
        consecutive_rejects = 0
        is_alarming = False # Reset alarm state on accepted part
    else:
        rejected_parts += 1
        consecutive_rejects += 1
        # Tool-wear alarm condition and counting logic (Do NOT change)
        if consecutive_rejects >= CONSECUTIVE_REJECTS_THRESHOLD:
            if not is_alarming: # Only increment alarm count for a *new* alarm event
                tool_wear_alarms += 1
                if first_alarm_part_number is None:
                    first_alarm_part_number = part_number
                is_alarming = True
            # Print alarm message only if an alarm condition is met
            print(f"⚠ TOOL WEAR SUSPECTED at part {part_number} (Size: {part_size:.2f} mm)")
        else: # Rejected, but not yet reached the consecutive reject threshold for an alarm
            is_alarming = False # Ensure alarm state is false if rejected part doesn't trigger alarm yet

print("\nPart inspection complete.\n")

# ==================================================
# JOB C Requirements: Stamped Shift Report
# ==================================================

# 2. Create the current date and time
now = datetime.datetime.now()

# 4, 5, 6. Calculations for the report
operator_1_name = "[Your Name]"
operator_2_name = "[Partner Name]"

total_parts = TOTAL_PARTS_INSPECTED
acceptance_percentage = round(accepted_parts / total_parts * 100, 1)
nominal_area = round(math.pi * (NOMINAL_DIAMETER/2)**2, 2)

# Determine first alarm display
first_alarm_display = f"Part {first_alarm_part_number}" if first_alarm_part_number is not None else "None"

# 3, 7. Print an official shift report (Final Report Format)
print("=" * 60)
print(f"                 SHIFT REPORT · {now.strftime('%d-%m-%Y %H:%M')}")
print("=" * 60)
print(f"Date & Time          : {now.strftime('%d-%m-%Y %H:%M')}")
print(f"Operator 1           : {operator_1_name}")
print(f"Operator 2           : {operator_2_name}")
print("-" * 60)
print("Production Summary")
print("-" * 60)
print(f"Total Parts          : {total_parts}")
print(f"Accepted             : {accepted_parts}")
print(f"Rejected             : {rejected_parts}")
print(f"Acceptance           : {acceptance_percentage} %")
print("-" * 60)
print("Inspection Details")
print("-" * 60)
print(f"Tolerance Band       : {LOWER_TOLERANCE:.2f} - {UPPER_TOLERANCE:.2f} mm")
print(f"Nominal Area         : {nominal_area:.2f} mm2")
print("-" * 60)
print("Tool-Wear Monitoring")
print("-" * 60)
print(f"Alarms Raised        : {tool_wear_alarms}")
print(f"First Alarm          : {first_alarm_display}")
print("=" * 60)
```
